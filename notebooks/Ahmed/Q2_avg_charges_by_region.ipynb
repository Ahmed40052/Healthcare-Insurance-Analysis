```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('/content/drive/MyDrive/Colab Projects/Healthcare-Insurance-Analysis/data/processed/insurance_clean.csv')
df.head()
```

```python
# Average charges by region, sorted descending
result = df.groupby('region')['charges'].mean().sort_values(ascending=False)
print(result)
```

```python
# Bar chart
plt.figure(figsize=(8, 5))
result.plot(kind='bar', color='#17a398', edgecolor='black')
plt.title('Average Medical Charges by Region')
plt.ylabel('Average Charges ($)')
plt.xlabel('Region')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
summary = df.groupby('region')['charges'].agg(['mean', 'count']).sort_values('mean', ascending=False)
summary.columns = ['avg_charges', 'num_patients']
summary['avg_charges'] = summary['avg_charges'].round(2)
print(summary)
```
