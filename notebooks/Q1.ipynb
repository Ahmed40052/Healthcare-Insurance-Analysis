```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("insurance_clean.csv")
```

```python
# 1. Average medical charges by smoking status
avg_by_smoker = df.groupby('smoker')['charges'].mean().round(2)

print("Average medical charges by smoking status: \n", avg_by_smoker)
```

```python
# 2. Average medical charges by region
avg_by_region = (
    df.groupby('region')['charges']
    .mean()
    .round(2)
    .sort_values(ascending=False)
)
print("\nAverage medical charges by region:\n", avg_by_region)
```

```python
# 3. Average medical charges by number of children
avg_by_children = df.groupby('children')['charges'].mean().round(2)

print("\nAverage medical charges by number of children:\n", avg_by_children)
```

```python
# 4. Combined analysis: smoking status + region + number of children
avg_combined = (
    df.groupby(['smoker', 'region', 'children'])['charges']
    .mean()
    .round(2)
)

print("\nAverage medical charges by smoking status, region, and number of children:\n")
print(avg_combined)
```

Finding:
- Smokers have much higher average medical charges than non-smokers.
- Southeast has the highest average charges by region.
- Medical charges vary across different numbers of children.
- Smoking shows a strong association with higher medical charges.
