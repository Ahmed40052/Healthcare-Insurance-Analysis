```python
import pandas as pd
import numpy as np

df = pd.read_csv('insurance_clean.csv')
print(df.shape)

df['bmi_category'] = pd.cut(
    df['bmi'],
    bins=[0, 18.5, 25, 30, float('inf')],
    labels=['Underweight', 'Normal', 'Overweight', 'Obese']
)
print(df['bmi_category'].value_counts())

# 1. BMI category
bmi_only = df.groupby('bmi_category')['charges'].agg(['mean', 'median', 'count']).round(2)
print("=== BMI Category Only ===")
print(bmi_only)
print()

# 2. Smoker
smoker_only = df.groupby('smoker')['charges'].agg(['mean', 'median', 'count']).round(2)
print("=== Smoker Status Only ===")
print(smoker_only)
print()

# 3. (BMI × Smoker)
combo = df.groupby(['bmi_category', 'smoker'])['charges'].agg(['mean', 'median', 'count']).round(2)
print("=== BMI Category × Smoker ===")
print(combo)

# Compute correlation between age, bmi, and charges
correlation = df[['age', 'bmi', 'charges']].corr()
print(correlation.round(3))

# Bucket age into groups for easier interpretation
df['age_group'] = pd.cut(df['age'], bins=[17, 29, 39, 49, 59, 100],
                          labels=['18-29', '30-39', '40-49', '50-59', '60+'])

# Compute average bmi and charges per age group
age_analysis = df.groupby('age_group')[['bmi', 'charges']].mean().round(2)
print(age_analysis)

# Calculate Q1, Q3, and IQR for charges
Q1 = df['charges'].quantile(0.25)
Q3 = df['charges'].quantile(0.75)
IQR = Q3 - Q1

# Define lower and upper bounds for outliers
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print(f"Q1: {Q1:.2f}, Q3: {Q3:.2f}, IQR: {IQR:.2f}")
print(f"Lower bound: {lower_bound:.2f}, Upper bound: {upper_bound:.2f}")

# Flag rows considered outliers
outliers = df[(df['charges'] < lower_bound) | (df['charges'] > upper_bound)]
print(f"\nNumber of outliers: {len(outliers)}")

# Check the composition of the outliers
print("\nSmoker distribution among outliers:")
print(outliers['smoker'].value_counts(normalize=True).round(3))

print("\nBMI category distribution among outliers:")
print(outliers['bmi_category'].value_counts(normalize=True).round(3))
```
