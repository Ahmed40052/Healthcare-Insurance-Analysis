```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sns.set_theme(style="whitegrid")
```

```python
df = pd.read_csv('insurance_clean.csv')
```

```python
# Calculate average charges per age to see the trend
avg_cost_age = df.groupby('age')['charges'].mean().reset_index()
```

```python
bins = [17, 25, 35, 45, 55, 65]
labels = ['18-25', '26-35', '36-45', '46-55', '56-65']
df['Age Group'] = pd.cut(df['age'], bins=bins, labels=labels)
avg_cost_age_group = df.groupby('Age Group')['charges'].mean().reset_index()
plt.figure(figsize=(8, 5))
sns.lineplot(data=avg_cost_age_group, x='Age Group', y='charges', color='purple', marker='o', linewidth=2)
plt.title('Average Medical Charges by Age Group')
plt.ylabel('Average Charges')
plt.xlabel('Age Group')
plt.show()
```
