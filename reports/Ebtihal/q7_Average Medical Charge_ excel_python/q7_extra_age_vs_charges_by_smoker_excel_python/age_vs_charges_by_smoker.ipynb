```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('/content/drive/MyDrive/Colab Projects/Healthcare-Insurance-Analysis/insurance.csv')
df.head()
```

```python
plt.figure(figsize=(9, 6))
sns.scatterplot(data=df, x='age', y='charges', hue='smoker',
                 palette={'yes': '#e07a5f', 'no': '#3d9970'}, alpha=0.7)
plt.title('Age vs. Charges, Distinguished by Smoker Status', fontsize=13, fontweight='bold')
plt.xlabel('Age')
plt.ylabel('Charges ($)')
plt.legend(title='Smoker')
plt.tight_layout()
plt.show()
```
