```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('insurance_clean.csv')
```

```python
# Calculate numerical averages
avg_cost_gender = df.groupby('sex')['charges'].mean()
print("Average Charges by Gender:\n", avg_cost_gender, "\n")

avg_cost_gender_smoker = df.groupby(['sex', 'smoker'])['charges'].mean().unstack()
print("Average Charges by Gender and Smoking Status:\n", avg_cost_gender_smoker)
```

```python
sns.barplot(data=df, x='sex', y='charges', hue='smoker', palette={'yes': '#FF9999', 'no': '#89CFF0'})
plt.title('Average Medical Charges by Gender and Smoking Status')
plt.ylabel('Average Charges')
plt.xlabel('Gender')
plt.show()
```
